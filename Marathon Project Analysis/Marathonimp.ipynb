```python
# lets fix the column names before fixing the nulls 

for columns in df_ind.columns:
    df_ind.rename(columns= {columns : columns.replace(" ","_").lower()},inplace= True)
```

```python
# df_ind is going to be the dummy table

ind_df = df_ind
```

```python
# Dropping na values in dummy table 

df_ind = df_ind.dropna()
```

```python
df_ind.isna().sum()
```

```python
# Group and calculate the median year of birth for each gender and event year
# This is done in order to fill na value in the year of birth column based on certain condition 

median_year = df_ind.groupby(['athlete_gender', 'year_of_event'])['athlete_year_of_birth'].median().reset_index()

def filling_year(data):
    if pd.isna(data['athlete_year_of_birth']):
        gender = data['athlete_gender']
        event_year = data['year_of_event']
        
        median_row = median_year[(median_year['athlete_gender'] == gender) & (median_year['year_of_event'] == event_year)]
        median_row_alt = median_year[median_year['year_of_event'] == event_year]
        median_row_9 = median_year[(median_year['athlete_gender'] == gender) & (median_year['year_of_event'] == 2009)]

        if not median_row.empty:
            return median_row['athlete_year_of_birth'].iloc[0]
        elif not median_row_alt.empty:
            return median_row_alt['athlete_year_of_birth'].iloc[0]
        elif (data['year_of_event'] == 2007 or data['year_of_event'] == 2008) and not median_row_9.empty:
            return median_row_9['athlete_year_of_birth'].iloc[0]
    return data['athlete_year_of_birth']

# Apply the function to fill the missing values
ind_df['athlete_year_of_birth'] = ind_df.apply(filling_year, axis=1)
```

```python
ind_df.isna().sum()
```

```python
# Only including Events held in 2000's

ind_df = ind_df[ind_df['year_of_event'] >= 2000]
```

```python
# Dropping Uneccecary Columns

ind_df.drop(columns= ['athlete_club','athlete_age_category'],inplace= True)
```

```python
# Rounding off athlete year of birth column 

ind_df['athlete_year_of_birth'] = ind_df['athlete_year_of_birth'].round()
```

```python
# Checking data types

ind_df.dtypes
```

```python
# Changing data types 

ind_df['athlete_year_of_birth'] = ind_df['athlete_year_of_birth'].astype(int)
ind_df['event_date'] = pd.to_datetime(ind_df['event_date'], format= 'mixed')
ind_df['athlete_average_speed'] = ind_df['athlete_average_speed'].astype(float)
ind_df['athlete_id'] = ind_df['athlete_id'].astype(object)
```

```python
# genrating athlete age

ind_df['athlete_age'] = ind_df['year_of_event'] - ind_df['athlete_year_of_birth']
```

```python
# fixing athlete performance column 

ind_df['athlete_performance(h)'] = ind_df['athlete_performance'].str.split(":").apply(lambda x : ".".join([i for i in x[0:2]]))
```

```python
# let only include 50's event_distance

ind_df50 = ind_df[ind_df['event_distance/length'].isin(['50km', '52km', '53km','52.2km','56km', '51km',
                                                          '54km', '55km', '50.2km'])]
```

```python
# Adding Age Category 
def filling_age_category(data):
    if data['athlete_age'] >= 14 and data['athlete_age'] <= 24:
        return "Youth\Young Athlete"
    elif data['athlete_age'] >= 25 and data['athlete_age'] <= 34:
        return "Prime Age Athletes"
    elif data['athlete_age'] >= 35 and data['athlete_age'] <= 39:
        return "Semi-Middel Age Athletes"
    elif data['athlete_age'] >= 40 and data['athlete_age'] <= 49:
        return "Middle-Aged Athletes"
    elif data['athlete_age'] >= 50 and data['athlete_age'] <= 59:
        return "Semi-Old Athletes"
    elif data['athlete_age'] >= 60 and data['athlete_age'] <= 69:
        return "Senior Athletes"
    elif data['athlete_age'] >= 70:
        return "Veteran Athletes"


ind_df50['age_category'] = ind_df50.apply(filling_age_category,axis= 1)
```

```python
# Fixing event_name 

ind_df50['event_name'] = ind_df50.apply(lambda row: f"{row['event_name']}-{row['year_of_event']}",axis= 1)
ind_df50.head()
```
